Делаем градиентный спуск с помощью NumPy без использования Sklearn.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
X = np.array([1, 2, 3])
y = np.array([2, 4, 6])
```

```python
def gradient_global(w, X, y):
    return sum(2 * (y - w * X) * -X) / len(y)
```

```python
def gradient_b(w, b, X, y):
    return sum(2 * (w * X + b - y)) / len(y)
```

```python
def mse(w, b, X, y):
    return sum((y - (X * w + b)) ** 2) / len(y)
```

```python
def gradient_descent(w, b, X, y, learning_rate, n_iterations):
    loss_history = []
    w_history = []
    b_history = []
    for _ in range(n_iterations):
        w_new = w - learning_rate * gradient_global(w, X, y)
        b_new = b - learning_rate * gradient_b(w, b, X, y)
        w = w_new
        b = b_new
        loss = mse(w, b, X, y)
        loss_history.append(loss)
        w_history.append(w)
        b_history.append(b)
    return w, b, loss_history, w_history, b_history
```

```python
w1, b1, loss_history1, w_hist1, b_hist1 = gradient_descent(0, 0, X, y, 0.01, 20)
w_hist1, b_hist1
```

```python
def new_mse(w, b, X, y):
    return sum((y - (X @ w + b)) ** 2) / len(y)
```

```python
def new_gradient_descent(w, b, X, y, learning_rate, n_iterations):
    loss_history = []
    w_history = []
    b_history = []
    for _ in range(n_iterations):
        y_pred = X @ w + b
        errors = y_pred - y

        gradient_w = 2 * (X.T @ errors) / len(y)
        gradient_b = 2 * np.sum(errors) / len(y)

        w_new = w - learning_rate * gradient_w
        b_new = b - learning_rate * gradient_b

        w = w_new
        b = b_new

        loss = new_mse(w, b, X, y)
        loss_history.append(loss)
        w_history.append(w)
        b_history.append(b)

    return w, b, loss_history, w_history, b_history
```
