use imputation to fill Nan

```python
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
```

```python
numerical_features = ["Pclass", "Age", "SibSp", "Parch", "Fare"]
categorical_features = ["Sex", "Embarked"]
```

```python
numerical_transformer = SimpleImputer(strategy="median")  # replace NaN with the median
```

```python
categorical_transformer = SimpleImputer(strategy="most_frequent")  # replace NaN with most frequent value
```

```python
categorical_encoder = OneHotEncoder(handle_unknown="ignore")  # encodes categorical features into numeric
# vectors and doesnt crash if unknown value is encountered in validation
```

```python
categorical_pipeline = Pipeline([("imputer", categorical_transformer), ("encoder", categorical_encoder)])
```

```python
preprocessor = ColumnTransformer(
    [
        ("numerical", numerical_transformer, numerical_features),
        ("categorical", categorical_pipeline, categorical_features),
    ]
)
# позволяет применить разные способы обработки к разным столбцам датасета
```

```python
model = RandomForestClassifier(n_estimators=100, random_state=42)
```

```python
model_pipeline = Pipeline([("preprocessor", preprocessor), ("model", model)])
```

Baseline-model

```python
model_pipeline.fit(X_train, y_train)
```

```python
predictions = model_pipeline.predict(X_val)
```

```python
def add_features(X):
    X = X.copy()
    X["familySize"] = X["Parch"] + X["SibSp"] + 1
    X["isalone"] = (X["familySize"] == 1).astype("int")
    X["Title"] = X["Name"].str.extract(r",\s*([^.]*)\.")

    title_mapping = {"Mlle": "Miss", "Mme": "Mrs", "Ms": "Mrs"}

    X["Title"] = X["Title"].replace(title_mapping)

    common_titles = ["Mr", "Miss", "Mrs", "Master"]
    X.loc[~X["Title"].isin(common_titles), "Title"] = "Rare"
    return X
```
