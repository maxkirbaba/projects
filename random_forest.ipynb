```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split
```

```python
melburn_file_path = "/home/den/max/projects/kaggle_datasets/melbourne-housing-snapshot/melb_data.csv"
melburn_data = pd.read_csv(melburn_file_path)

melburn_data = melburn_data.dropna(axis=0)

y = melburn_data.Price
melbourne_features = ["Rooms", "Bathroom", "Landsize", "BuildingArea", "YearBuilt", "Lattitude", "Longtitude"]

X = melburn_data[melbourne_features]

train_X, val_X, train_y, val_y = train_test_split(X, y, random_state=0)
```

```python
model = RandomForestRegressor(random_state=1)
model.fit(train_X, train_y)
preds_val = model.predict(val_X)
mae = mean_absolute_error(preds_val, val_y)
mae
```
